```python
from scipy import stats
import numpy as np
```

```python
# Chi squared - Test of Independence

'''
Gender and product preference

Null hypothesis:
Gender and product preference are independent.

Alternative hypothesis:
Gender and product preference are not independent.
'''

table = np.array([[60, 40],[50, 50]])

Chi2, Pval, dof, result = stats.chi2_contingency(table)

print(f"Chi-square value : {Chi2}")
print(f"P value          : {Pval}")
print(f"Degrees of freedom : {dof}")

print(result)

if Pval < 0.05:
    print("\nAlternative hypothesis is considered, so null hypothesis is rejected.")
else:
    print("\nNull hypothesis is not rejected. Gender and Product preference are independent")
```
