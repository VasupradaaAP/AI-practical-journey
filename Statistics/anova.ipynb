```python
from scipy import stats
import numpy as np
```

```python
# Anova - One way

'''
Average productivity of three training programs

Null hypothesis:
Mean(A) = Mean(B) = Mean(C)

Alternative hypothesis:
At least one program has a different mean.
'''

program_A = np.random.normal(loc=80, scale=8, size=25)
program_B = np.random.normal(loc=75, scale=7, size=30)
program_C = np.random.normal(loc=82, scale=6, size=28)

Fval, Pval = stats.f_oneway(program_A, program_B, program_C)
print(f"F value: {Fval}")
print(f"P value: {Pval}")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected.")
else:
    print("Null hypothesis is not rejected.")
```
