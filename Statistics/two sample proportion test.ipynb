```python
from scipy import stats
import numpy as np
```

```python
# Two sample - Proportion test - 2 tail

'''
Different versions of websites are tested 
Website A : 318 / 500
Website B : 290 / 480

null hypothesis : A = B
alter hypothesis : A > B (or) A < B
'''

from statsmodels.stats.proportion import proportions_ztest

Zval, Pval = proportions_ztest(count=[318,290] ,nobs=[500,480], alternative="two-sided")
print(f"T value: {Zval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected, There is no significant difference between Website Version A and B")
```

```python
# Two sample - Proportion test - 2 tail

'''
Vaccines are tested in two regions
Region A : 180 / 300
Region B : 165 / 310

null hypothesis : A = B
alter hypothesis : A > B (or) A < B
'''

from statsmodels.stats.proportion import proportions_ztest

Zval, Pval = proportions_ztest(count=[180,165] ,nobs=[300,310], alternative="two-sided")
print(f"T value: {Zval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected, There is no significant difference between Vaccine A and B")
```
