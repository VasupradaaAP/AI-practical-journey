```python
from scipy import stats
import numpy as np
```

```python
# Chi squared - Goodness of fit

'''
Dice is fair or not

Null hypothesis:
The die is fair.

Alternative hypothesis:
The die is not fair.
'''

observed = [18, 22, 20, 25, 17, 18]
expected = [20, 20, 20, 20, 20, 20]

Chi2, Pval = stats.chisquare(f_obs = observed, f_exp = expected)

print("Chi-square value: {:.3f}".format(Chi2))
print(f"P value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected.")
    print("The die is not fair.")
else:
    print("Null hypothesis is not rejected.")
    print("There is no significant difference that the die is unfair.")
```
