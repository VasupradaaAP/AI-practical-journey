```python
from scipy import stats
import numpy as np
```

```python
# One sample - Proportion test - 1 tail

'''
Avg people are satisfied with the product
null hypothesis : <= 60%
alter hypothesis : > 60% [Avg: 67%]
'''

from statsmodels.stats.proportion import proportions_ztest

Tval, Pval = proportions_ztest(count=134 ,nobs=180, value=0.60, alternative="larger")
print(f"T value: {Tval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
```

```python
# One sample - Proportion test - 1 tail

'''
Avg applicants pass the exam
null hypothesis : <= 40%
alter hypothesis : > 40% [Avg: 50%]
'''

from statsmodels.stats.proportion import proportions_ztest

Tval, Pval = proportions_ztest(count=90 ,nobs=180, value=0.40, alternative="larger")
print(f"T value: {Tval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
```
