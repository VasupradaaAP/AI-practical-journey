```python
from scipy import stats
import numpy as np
```

```python
# One sample - T test - 1 tail

'''
Avg score taken by students
null hypothesis : <=65
alter hypothesis : >65 [Sample: 50, Avg:68 with SD: 10]
'''
score = np.random.normal(loc=68, scale=10, size=50).astype(int)
print(f"{score}\n")

Tval, Pval = stats.ttest_1samp(score, popmean=65, alternative="greater")
print(f"T value: {Tval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
```

```python
# One sample - T test - 1 tail

'''
Avg time taken to resolve customer
null hypothesis : <=30
alter hypothesis : >30 [Avg: 33 with SD: 8 mins]
'''
tickets = np.random.normal(loc=33, scale=8, size=40).astype(int)
print(f"{tickets}\n")

Tval, Pval = stats.ttest_1samp(tickets, popmean=30, alternative="greater")
print(f"T value: {Tval}\nP value: {Pval}\n")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected")
```
