```python
from scipy import stats
import numpy as np
```

```python
# Two sample - T test - 2 tail

'''
Avg scores between 2 teaching methods
Method A mean is 75 with SD 12
Method B mean is 70 with SD 10

null hypothesis : A = B
alter hypothesis : A > B (or) A < B
'''

method_a = np.random.normal(loc=75, scale=12, size=30).astype(int)
method_b = np.random.normal(loc=70, scale=10, size=28).astype(int)
#print(a)
#print(b)

#Tval, Pval = stats.ttest_ind(method_a, method_b, alternative="two-sided")
#print(f"T value: {Tval}\nP value: {Pval}\n")

Tval, Pval = stats.ttest_ind_from_stats(mean1=75, std1=12, nobs1=30,
                                        mean2=70, std2=10, nobs2=28,
                                        alternative="two-sided")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected. There is no significant difference between Method A and B")
```

```python
# Two sample - T test - 2 tail

'''
Difference in mean delivery time between warehouses
Warehouse A mean is 48 with SD 6
Warehouse B mean is 52 with SD 7

null hypothesis : A = B
alter hypothesis : A > B (or) A < B
'''

wh_a = np.random.normal(loc=48, scale=6, size=35).astype(int)
wh_b = np.random.normal(loc=52, scale=7, size=40).astype(int)
#print(a)
#print(b)

#Tval, Pval = stats.ttest_ind(wh_a, wh_b, alternative="two-sided")
#print(f"T value: {Tval}\nP value: {Pval}\n")

Tval, Pval = stats.ttest_ind_from_stats(mean1=48, std1=6, nobs1=35,
                                        mean2=52, std2=7, nobs2=40,
                                        alternative="two-sided")

if Pval < 0.05:
    print("Alternative hypothesis is considered, so null hypothesis is rejected")
else:
    print("Null hypothesis is not rejected. There is no significant difference between Method A and B")
```
