```python
import pandas as pd

emp1 = pd.read_csv("dataset/employee_data1.csv")
emp2 = pd.read_csv("dataset/employee_data2.csv")
emp1.head()
```

```python
emp1.groupby(by="Department")["Salary"].sum().sort_values()
```

```python
pd.pivot_table(data=emp1, index="Department", values="Salary", aggfunc=sum).sort_values(by="Salary", ascending=False)
```

```python
emp1.groupby(by="Department")["Salary"].agg(func=['min','max','sum','mean']).round(2)
```

```python
merged_data = pd.merge(emp1,emp2,how="outer")
merged_data.head()
```

```python
pd.crosstab(index = merged_data['Department'], columns = merged_data['Gender'], margins=True)
```

```python
merged_data.head(3)
```

```python
merged_data["Health Checkup"] = merged_data["Age"].apply(lambda a: "Mandatory" if a>30 else "Not mandatory")
```

```python
def games(dept):
    if dept == "IT" or dept == "IT Tech":
        return "Handball"
    else:
        return "Badminton"

merged_data["Game"] = merged_data["Department"].apply(games)
merged_data.head(2)
```

```python
merged_data.rename(columns={"Game":"Event Games"}, inplace=True)
merged_data.head(2)
```

```python
merged_data.columns = merged_data.columns.str.replace("Event Games","Events")
merged_data.head(2)
```
