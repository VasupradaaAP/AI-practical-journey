### Task 1
Data Cleaning Challenge on Sales Data

```python
import pandas as pd

data_2017 = pd.read_csv("dataset/sales_2017.csv")
data_2018 = pd.read_csv("dataset/sales_2018.csv")
data_2019 = pd.read_csv("dataset/sales_2019.csv")

all_sales = pd.concat(objs = [data_2017, data_2018, data_2019])
all_sales

all_sales.shape
```

```python
del all_sales["Voucher Amount"]
del all_sales["Gross"]
del all_sales["Disc"]
```

```python
all_sales.dtypes
```

```python
all_sales = all_sales.apply(lambda col: col.str.strip())
all_sales = all_sales.replace(r'^\s*$', pd.NA, regex=True, inplace=True)

all_sales.dropna(inplace=True)
all_sales.shape
```

```python
# Change in Data Type

all_sales['Date'] = pd.to_datetime(all_sales['Date'], format = '%d/%m/%Y')

all_sales['Voucher'] = all_sales['Voucher'].str.replace("Sal:","").astype(float).astype(int)

all_sales['Rate'] = all_sales['Rate'].str.replace(',',"").astype(float)

all_sales['Qty'] = all_sales['Qty'].str.replace(',',"").astype(float).astype(int)

all_sales[['Party', 'Product']] = all_sales[['Party', 'Product']].astype(object)
```

```python
all_sales.dtypes
```

```python
all_sales.duplicated().sum()
```

```python
all_sales.drop_duplicates(inplace=True)
```

```python
all_sales.shape
```

```python
print(all_sales.isnull().sum())
```

```python
print(all_sales.duplicated().sum())
```

```python
all_sales.head(10)
```

```python
all_sales = all_sales.to_csv("updateSales.csv", index=False)
```

Merge Function 1 Example

```python
# merge function

import pandas as pd

emp1 = pd.read_csv("dataset/employee_data1.csv")
emp2 = pd.read_csv("dataset/employee_data2.csv")

combine = emp1.merge(emp2, how="outer")
combine
```

```python
combine = emp1.merge(emp2, how="inner")
combine
```

```python
employee_datum = emp1.merge(emp2, how="right")
employee_datum
```

```python
employee_datum = emp1.merge(emp2).sort_values("EmployeeID")
employee_datum = employee_datum.reindex(columns=['EmployeeID', 'Name', 'Age', 'Gender', 'Department', 'Salary'])
employee_datum
```

### Task 3
List Comprehension 2 examples

```python
# Example 1

coin = [(i,j) for i in ['H','T'] for j in ['H','T']]
print(coin)
```

```python
# Example2

word = input("Enter the name: ")

condition_1 = [i for i in word if i not in 'aeiouAEIOU']
print(f"Word '{word}' consonants are - *{''.join(condition_1)}*")

condition_2 = [i if i not in 'aeiouAEIOU' else '_' for i in word]
print(f"Word : {''.join(condition_2)}")
```

### Task 4
Lambda Function 2 examples with Map, Filter, Reduce

```python
# Example 1 - map

list_1 = [1,2,3,4,5]
list_2 = [6,7,8,1,2]

result = map(lambda x, y: x + y, list_1, list_2)
print(list(result))
```

```python
# Example 2 - map

demo = [50,45,60,70,80,95]

operation1 = map(lambda answer: answer/5, demo)
print(list(operation1))                         # using map and lambda


def div(a):
    return a/10
operation2 = map(div,demo)

print(list(operation2))                         # using only map
```

```python
# Example 1 - filter

number_filter = "You know! The *Report*, published in 2025, contains information about ancient sculptures dating back to around 2000 BC."
l = list(filter(lambda a: a == '*' or a == ',' or a == '!' or a == '.' or a >= '0' and a <= '9', number_filter))
print(f"Length : {len(l)}, {l}")
```

```python
# Example 2 - filter

Age = [34,65,78,24,17,42,67,14,10,45,9,2,34]
list(filter(lambda a:a >= 18, Age))
```

```python
# Example 1 - reduce

from functools import reduce

data = ['p','y','t','h','o','n']
mul = lambda m, d : m + d
reduce(mul,data)
```

```python
# Example 2 - reduce

from functools import reduce

fibonacci = [0,1,1,2,3,5,8,13,21]

mul = lambda m, fact : m + fact
reduce(mul,fibonacci)
```
