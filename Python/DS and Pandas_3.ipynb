```python
import seaborn as sb
import pandas as pd
import matplotlib.pyplot as mp

datum = pd.read_csv("footwear.csv")
datum.head()
```

```python
del datum['Category']
```

```python
datum['Product Type'].unique()
```

```python
datum['Product Type'].value_counts()
```

```python
datum.head()
```

```python
mp.pie(data = datum, 
       x = datum["Product Type"].value_counts(),
       labels = ['T-shirt', 'Sneakers', 'Hoodie', 'Cap', 'Joggers'],
       autopct = "%.2f%%", explode=[0.01,0.01,0.02,0.02,0.02])

mp.show()
```

```python
sb.barplot(datum, x='Country' , y='Amount ($)')
mp.show()
```

```python
sb.barplot(datum, x='Product Type' , y='Amount ($)')
mp.show()
```

```python
datum.dtypes
```

```python
sb.heatmap(datum.select_dtypes(include='number').corr().round(2), annot=True)
mp.show()
```

```python
sb.histplot(datum, x="Unit Price ($)", hue="Gender" ,multiple="stack")
mp.show()
```

```python
sb.countplot(data=datum, x='Country', hue='Product Type')
mp.title("Products sold by different countries")
mp.xlabel("COUNTRY")
mp.ylabel("PRODUCT COUNT")
mp.show()
```

```python
adds = pd.crosstab(
    index=datum['Country'],
    columns=datum['Product Type'],
    values=datum['Quantity'],
    aggfunc='sum')


sb.heatmap(adds, annot=True)
mp.show()
```
