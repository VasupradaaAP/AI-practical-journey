## Duplicates

```python
import pandas as pd
sport = pd.read_csv("sports.csv")
sport
```

```python
sport[sport.duplicated(keep=False)]
```

```python
sport[sport.duplicated(keep=False)]

sport.drop_duplicates(inplace=True)
sport
```

```python
# sport.info()
sport.dtypes
#sport["Fees"].dtypes
#type(sport)                # pandas.DataFrame
#type(sport["Fees"])        # pandas.Series
#type(sport["Name"].values) # pandas.arrays.StringArray
```

```python
sport.dropna(subset=["Fees"], inplace=True)
sport
```

```python
try:
    sport["Fees"] = sport["Fees"].str.replace("Rs.", "").astype(float).astype(int)
except:
    pass
sport
```

```python
sport["Sport"] = sport["Sport"].fillna("Tennis")
sport
```

```python
# Specific row and column value update

sport.loc[sport["Name"] == "Mani", "Sport"] = "Tenikoit"    # sport.loc[7, 'Sport'] = 45
sport.loc[sport["Name"] == "Kumar", "Sport"] = "Badminton"
sport
```
