## Task 1
Give 2 examples of all String functions which should be meaningful

```python
# Example 1

string = "How to track the content of AI Technology in 2026 with all the available resources?"

print(string.capitalize())
print(string.upper())
print(string.title())
print(string.replace("AI", "Artificial Intelligence"))
print(len(string))
print(string.find("track"))
print(string.swapcase())
```

```python
# Example 2

string = "Make use of the resources which you have, so that you can finish your work on time and with quality."

print(string.isalpha())
print(string.isdigit())
print(string.split())
print(len(string.split()))
print(string.endswith("."))
print(string.lower())
print(string.count("t"))

```
