### Data Visualization ###

```python
import matplotlib.pyplot as mp
import seaborn as sb
import pandas as pd
```

```python
carData = pd.read_csv("cars.csv")
```

```python
carData.isnull().sum()
carData.duplicated().sum()
```

```python
carData.head()
```

```python
mp.hist(carData['disp'])
mp.xlabel("displacement")
mp.ylabel("freq")
mp.title("Cars information")
mp.show()
```

```python
carData.model.unique()
```

```python
spdata=pd.read_csv("sports.csv")
sb.countplot(data=spdata, x='Sport')
```

```python
spdata.Sport.value_counts()
```

```python
sb.boxplot(carData['hp'])
```

```python
mp.plot(spdata['Location'])
```

```python
sb.barplot(data=carData, x='cyl', y='drat')
```

```python
sb.scatterplot(data=carData, x='disp', y='qsec')
```
