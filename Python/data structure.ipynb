```python
indian_states.insert(3, "Goa")
print(indian_states)
```

## CREATE THREE DataFrame USING DICT

```python
shapes = {"Square":[4,"cuboid","yellow"],
          "Circle":[0,"sphere","blue"],
          "Triangle":[3,"pyramid","orange"]}
print(shapes)

shop = {"Biscuits":["ParleG",10,120],
        "Chocolates":["DairyMilk",5,200],
        "Candies":["Snickers",15,300]}
print(shop)

dept = {"CSE":["Data structures","Operating Systems","Computer Networks"],
        "ECE":["Digital Electronics","Analog Electronics","Microprocessors"],
        "MECH":["Thermodynamics","Fluid Mechanics","Manufacturing Processes"]}
print(dept)
```

```python
import pandas as pd

display1 = pd.DataFrame(shapes, index = ["Sides","3D Shape","Colour"])
print(display1,"\n")

display2 = pd.DataFrame(shop, index = ["Product","Quantity","Price"])
print(display2,"\n")

display3 = pd.DataFrame(dept, index = ["Subject 1","Subject 2","Subject 3"])
print(display3)
```
