### Abstraction ###

```python
from abc import ABC, abstractmethod

class AmazonOrder(ABC):
    @abstractmethod
    def payment(self):
        pass

class NetBanking(AmazonOrder):
    def payment(self):
        return f"Net banking payment was successful!"

class CreditCard(AmazonOrder):
    def payment(self):
        return f"Credit Card payment was successful!"

pay1 = NetBanking()
pay2 = CreditCard()

pay1.payment()
pay2.payment()
```

### Encapsulation

```python
class amazon_acc:

    def __init__(self, name, pw):
        self.name = name
        self.__pw = pw

    def login(self):
        if self.__pw == "aar567":
            return f"Welcome {self.name.capitalize()}. Book your order !"
        else:
            return f"Check your password! Try again. Your password is {self.pw}"
        
account = amazon_acc("aarav","aar567")
account.login()
```

### Polymorphism

```python
class AirColler:
    def order(self):
        return f"Air Coller - ordered."
    
class Book:
    def order(self):
        return f"Book - ordered."
    
class Headphones:
    def order(self):
        return f"Headphones - ordered."
    
class Bag:
    def order(self):
        return f"Bag - ordered."
    

order_items = [AirColler(), Book(), Headphones(), Bag()]
for k in order_items:
    print(k.order())

```

### Inheritance

##### Single


```python
class Cart:

    def add_cart(self):
        if self.f == 1:
            return f"{self.prod} - added to cart"
        else:
            return f""
    
class Product(Cart):

    def __init__(self,prod,f=0):
        self.prod = prod
        self.f = f

    def view_prod(self):
        products = ["cotton","headset","sunglass","shirt"]

        if self.prod in products:
            self.f = 1
            return f"The product '{self.prod}' is available"
        else:
            return f"Unavailable product - {self.prod}"
    
    
buy = Product("headset")
print(buy.view_prod())
print(buy.add_cart())
```

##### Multiple

```python
class login:
    def Login(self):
        return f"\nExplore the Amazon products"
    
class cart:
    def Cart(self):
        return f"Selected! Product added to cart"
    
class buy(login, cart):
    def Buy(self, get):
        if get == 1:
            return f"\nProceed to buy ..."
        else:
            return self.Login()
    
proceed = buy()
print(proceed.Login())
print(proceed.Cart())
print(proceed.Buy(1))
```

##### Multi Level

```python
class ordered:
    def Ordered(self):
        return f"Ordered Product"
    
class track_order(ordered):
    def track(self):
        return f"Track the order ..."
    
class wrong_item(track_order):
    def wrong(self):
        return f"Received wrong item"
    
class return_order(wrong_item):
    def Return(self, ch):
        if ch.upper() == 'Y':
            return f"Request is sent to return the item. Thank you"
        else:
            return f"Sorry for the inconvenience"
    

cancel = return_order()
print(f"{cancel.Ordered()} \n{cancel.track()} \n{cancel.wrong()}")
ch = input()
print(f"{cancel.Return(ch)}")
```

##### Hierarchical



```python
class make_payment:
    def doPay(self):
        return f"Redirected to payment page ..."
    
class hardcopy(make_payment):
    def offlineBook(self):
        return f"Hardcopy price - Rs.200"
    
class kindle(make_payment):
    def onlineBook(self):
        return f"Amazon Kindle price - Rs.135"
    
buy1 = hardcopy()
print(f"{buy1.offlineBook()}, {buy1.doPay()}")

buy2 = kindle()
print(f"{buy2.onlineBook()}, {buy2.doPay()}")
```

##### Hybrid

```python
class orders_display:
    def display(self):
        print("Your list of ORDERS")

class received:
    def receive(self):
        print("Product received")

class message(received):
    def mess(self):
        print("Acknowledged through message - User")

class feedback(message):
    def feed(self):
        print("Provided feedback - User")

class goto_orderpage(feedback, orders_display):
    def view_orders(self):
        print("Moving to orders page ...")

obj = goto_orderpage()

obj.receive()
obj.mess()
obj.feed()
obj.view_orders()
obj.display()
```
